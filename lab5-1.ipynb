## Librerias

```python
import numpy as np
import matplotlib.pyplot as plt 
```

## Funcion

```python
def F(x, y, f):
    dx = 1
    dy = f(x, y)
    return dx, dy


def plot_meshgrid(  
                  xmin=-5, 
                  xmax=5, 
                  ymin=-5, 
                  ymax=5, 
                  xstep=1, 
                  ystep=1,
                  f=None):
    x = np.arange(xmin, xmax + xstep, xstep)
    y = np.arange(ymin, ymax + ystep, ystep)
    X, Y = np.meshgrid(x, y)
    DX, DY = F(X, Y, f)
    norm = np.sqrt(DX**2 + DY**2) 
    DX, DY = DX/norm, DY/norm    
    plt.quiver(X, Y, DX, DY, color="purple")
    plt.xlim(xmin, xmax)
    plt.ylim(ymin, ymax)
    plt.gca().set_aspect("equal")
    plt.show()
    
    
```

### Ejemplo 1

```python
f = lambda x, y: x - y 
plot_meshgrid(f=f, xmin=-5, xmax=5, ymin=-5, ymax=5, xstep=1, ystep=1)
```

### Ejemplo 2

```python
f2 = lambda x, y: np.sin(x) + np.cos(y)
plot_meshgrid(f=f2, xmin=-5, xmax=5, ymin=-5, ymax=5, xstep=0.5, ystep=0.5)
```
