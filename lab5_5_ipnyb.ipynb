## a) análisis dimensional
t está en semanas y P en individuos. dP/dt tiene unidades de individuos por semana. Para que 0.06·P tenga esas mismas unidades, 0.06 debe ser 1/semana. Para que 0.0004·P² tenga esas mismas unidades, 0.0004 debe ser 1/(individuo·semana). Queda: \[0.0004] = individuo⁻¹·semana⁻¹ y \[0.06] = semana⁻¹.

## b) puntos de equilibrio y estabilidad
la ecuación es P' = 0.0004 P² − 0.06 P = 0.0004 P(P − 150). Los equilibrios son P = 0 y P = 150. Con F(P)=0.0004 P(P−150), se tiene F'(P)=0.0008 P − 0.06. En P=0, F'(0) = −0.06 < 0 → estable (atrae). En P=150, F'(150) = 0.06 > 0 → inestable (repulsa). No aparecen semi-estables.

## c) esbozo cualitativo y concavidad
por el signo de P' = 0.0004 P(P−150): si 0 < P < 150 la población decrece hacia 0; si P > 150 la población crece y el término cuadrático domina, así que el modelo predice explosión en tiempo finito; en P = 0 y P = 150 las soluciones son constantes. Para la curvatura, uso P'' = F'(P)·F(P). El cambio de concavidad ocurre donde F'(P)=0, es decir en P = 75. Entonces: en 0 < P < 75 se tiene P''>0 (trayectorias cóncavas hacia arriba mientras bajan); en 75 < P < 150 se tiene P''<0 (cóncavas hacia abajo acercándose a 0); para P > 150 se tiene P''>0 (cóncavas hacia arriba mientras crecen). No hay máximos ni mínimos locales en las trayectorias no triviales, porque P' sólo se anula en los equilibrios.
