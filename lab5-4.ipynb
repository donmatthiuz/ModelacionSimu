```python
import numpy as np
import matplotlib.pyplot as plt
def F(x, y, f):
    dx = 1
    dy = f(x, y)
    return dx, dy


def plot_meshgrid(  
                  xmin=-5, 
                  xmax=5, 
                  ymin=-5, 
                  ymax=5, 
                  xstep=1, 
                  ystep=1,
                  f=None):
    x = np.arange(xmin, xmax + xstep, xstep)
    y = np.arange(ymin, ymax + ystep, ystep)
    X, Y = np.meshgrid(x, y)
    DX, DY = F(X, Y, f)
    norm = np.sqrt(DX**2 + DY**2) 
    DX, DY = DX/norm, DY/norm    
    plt.quiver(X, Y, DX, DY, color="purple")
    plt.xlim(xmin, xmax)
    plt.ylim(ymin, ymax)
    plt.gca().set_aspect("equal")
    plt.show()
    
    
```

## a) Campo de direcciones

```python



f = lambda x, y: (x - 3*y - 3*(x**2 - y**2) + 3*x*y) / (2*x - y + 3*(x**2 - y**2) + 2*x*y)


# Condición inicial
x0, y0 = 1.5, 0

# Rango y paso
xmin, xmax = -5, 5
dx = 0.01

# Crear arreglo de x
X = np.arange(x0, xmax + dx, dx)
Y = np.zeros_like(X)
Y[0] = y0

# Método de Euler
for i in range(1, len(X)):
    Y[i] = Y[i-1] + f(X[i-1], Y[i-1]) * dx


plot_meshgrid(f=f, xmin=-5, xmax=5, ymin=-5, ymax=5, xstep=0.5, ystep=0.5)

```

## b) Solucion del problema de valor inicial

```python
plt.plot(X, Y, color="red", linewidth=2, label=f"y({x0})={y0}")
plt.legend()
plt.show()
```
